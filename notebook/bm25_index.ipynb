```python
import pandas as pd
from rank_bm25 import BM25Okapi
import string
```

```python
df = pd.read_csv("../data/processed/dataset.csv")
# df = pd.read_csv("../docs_benchmark.csv")
docs = df['Descriptions']
```

```python
def removed_puncts(input_string):
    return input_string.translate(str.maketrans('','',string.punctuation)).lower()
```

```python
clearn_corpus = []
for doc in docs:
    clean_doc = removed_puncts(doc)
    clearn_corpus.append(clean_doc)
```

```python
tokenized_clean_corpus = []
for doc in clearn_corpus:
    doc = doc.split()
    tokenized_clean_corpus.append(doc)
```

```python
bm25 = BM25Okapi(tokenized_clean_corpus)
```

```python
query = "Phu Quoc"
tokenized_query = removed_puncts(query).split(" ")
bm25.get_top_n(tokenized_query, docs, n = 1)
```
